```python
import numpy as np
```

```python
offsets: np.ndarray = np.array(
    [[i,j] for i in range(-1,2) 
     for j in range(-1,2)
     if [i,j] != [0,0]])
offsets
```

```python
neighborCoords: np.ndarray = np.array([4, 2]) + offsets
neighborCoords
```

```python
neighborCoords[:,0]
```

```python
upper = 5
```

```python
rows = neighborCoords[:,0]
cols = neighborCoords[:,1]
mask = (rows >= 0) & (rows < upper) & (cols >= 0) & (cols < upper)
filtered = neighborCoords[mask]
filtered
```
